```python
import ast

def convert(text):
    L = []
    # Case 1: text is a string that looks like a list/dict
    if isinstance(text, str):
        try:
            items = ast.literal_eval(text)
        except:
            return []
    else:
        # Case 2: text is already a list
        items = text
    
    # Extract 'name' if dicts, otherwise keep the value
    for i in items:
        if isinstance(i, dict) and 'name' in i:
            L.append(i['name'])
        else:
            L.append(i)
    return L

# Apply safely
movie['genres'] = movie['genres'].apply(convert)
```

```python
import ast

def fetch_director(text):
    # Case 1: text is a string → parse it
    if isinstance(text, str):
        try:
            items = ast.literal_eval(text)
        except:
            return []
    else:
        # Case 2: text is already a list
        items = text
    
    # Extract director(s)
    directors = []
    for i in items:
        if isinstance(i, dict) and i.get('job') == 'Director':
            directors.append(i['name'])
        elif isinstance(i, str):
            # Already a string like 'James Cameron'
            directors.append(i)
    return directors

# Apply safely
movie['crew'] = movie['crew'].apply(fetch_director)
```

```python
def collapse(L):
    L1 = []
    for i in L:
        L1.append(i.replace(" ",""))
    return L1
```

```python
movie['cast'] = movie['cast'].apply(collapse)
movie['crew'] = movie['crew'].apply(collapse)
movie['genres'] = movie['genres'].apply(collapse)
movie['keywords'] = movie['keywords'].apply(collapse)
```

```python
movie.head()
```

```python
movie['overview'] = movie['overview'].apply(lambda x:x.split())
```

```python
movie['tags'] = movie['overview'] + movie['genres'] + movie['keywords'] + movie['cast'] + movie['crew']
```

```python
new = movie.drop(columns=['overview','genres','keywords','cast','crew'])
```

```python
new['tags'] = new['tags'].apply(lambda x: " ".join(x))
new.head()
```

```python
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer(max_features=5000,stop_words='english')
    
```

```python
vector = cv.fit_transform(new['tags']).toarray()
```

```python
vector.shape
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
similarity = cosine_similarity(vector)
```

```python
similarity
```

```python
new[new['title'] == 'The Lego Movie'].index[0]
```

```python
def recommend(movie):
    index = new[new['title'] == movie].index[0]
    distances = sorted(list(enumerate(similarity[index])),reverse=True,key = lambda x: x[1])
    for i in distances[1:6]:
        print(new.iloc[i[0]].title)
        
    
```

```python
print(recommend("Gandhi"))
```
